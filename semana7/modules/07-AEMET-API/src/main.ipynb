```python

import os

import requests
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("API_KEY")


# Consultar el inventario para localizar la estación
url_estaciones = (
    "https://opendata.aemet.es/opendata/api/valores/climatologicos/inventarioestaciones/todasestaciones/"
)

respuesta = requests.get(
    url_estaciones,
    params={"api_key": API_KEY},
    timeout=30
)
respuesta.raise_for_status()

# AEMET devuelve una URL en el campo "datos"
url_datos_estaciones = respuesta.json()["datos"]

estaciones = requests.get(url_datos_estaciones, timeout=30)
estaciones.raise_for_status()

df_estaciones = pd.DataFrame(estaciones.json())

# Buscar estaciones cuyo nombre contenga "Ciudad Universitaria"
print(
    df_estaciones[
        df_estaciones["nombre"].str.contains(
            "Ciudad Universitaria",
            case=False,
            na=False
        )
    ]
)
```
